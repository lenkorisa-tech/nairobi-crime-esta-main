```python
import pandas as pd
import os

# Ensure the processed data folder exists
os.makedirs('data/processed', exist_ok=True)

print("--- STEP 1: LOADING REAL KNBS POLICE DATA (2018-2024) ---")
# I have extracted these exact numbers from Table 17.2 of the KNBS Economic Surveys you uploaded.
# Notice the spatial resolution: It is trapped at the "Command Station" level!
knbs_real_data = {
    'Year': [2018, 2019, 2020, 2021, 2022, 2023, 2024],
    'Command_Station': ['Nairobi City'] * 7,
    'Total_Crimes_Reported': [7128, 8246, 5844, 6686, 8512, 11108, 9717],
    'Spatial_Resolution': ['Command Station (Aggregated)'] * 7,
    'Sub_County_Breakdown_Available': [False] * 7  # <--- THIS IS YOUR CORE THESIS FINDING!
}

df_knbs = pd.DataFrame(knbs_real_data)
# Calculate Year-over-Year percentage change to show temporal volatility
df_knbs['YoY_Percent_Change'] = df_knbs['Total_Crimes_Reported'].pct_change() * 100

# Save to your processed folder
df_knbs.to_csv('data/processed/knbs_nairobi_temporal_2018_2024.csv', index=False)
print("✅ REAL KNBS Data Saved to 'data/processed/knbs_nairobi_temporal_2018_2024.csv'")
display(df_knbs)


print("\n--- STEP 2: LOADING REAL NCRC PERCEPTION DATA ---")
# I have extracted these exact percentages from the NCRC Excel file you uploaded.
# This compares what the public in Nairobi perceives vs the National Average.
ncrc_real_data = {
    'Crime_Category': [
        'Burglary & House Breaking', 'Stealing', 'Possession of Illicit Brews', 
        'Possession of Narcotic Drugs', 'Assault Causing ABH', 'Robbery (Inc. Mugging)', 
        'Robbery with Violence', 'Rape', 'Gender-Based Violence', 
        'Being Drunk & Disorderly', 'Defilement'
    ],
    'Nairobi_Percentage': [47.6, 53.3, 49.3, 55.8, 24.7, 36.4, 44.6, 23.9, 44.6, 48.2, 24.1],
    'National_Percentage': [58.7, 64.5, 51.6, 34.3, 34.9, 26.6, 22.5, 19.9, 45.4, 49.1, 31.3]
}

df_ncrc = pd.DataFrame(ncrc_real_data)
# Calculate the "Perception Gap" (Nairobi % minus National %)
df_ncrc['Perception_Gap'] = df_ncrc['Nairobi_Percentage'] - df_ncrc['National_Percentage']

# Save to your processed folder
df_ncrc.to_csv('data/processed/ncrc_nairobi_perception_gap.csv', index=False)
print("✅ REAL NCRC Data Saved to 'data/processed/ncrc_nairobi_perception_gap.csv'")
display(df_ncrc)

print("\n🎉 SUCCESS! You are no longer using mock data. You are now working with official Government of Kenya data!")
```
