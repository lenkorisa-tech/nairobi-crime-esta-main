```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set visual style
sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (10, 6)

# 1. Load the KNBS data we just extracted
df_knbs = pd.read_csv('data/processed/knbs_nairobi_temporal_2018_2024.csv')

# 2. Create the Temporal Trend Chart
plt.figure(figsize=(10, 6))
sns.lineplot(data=df_knbs, x='Year', y='Total_Crimes_Reported', marker='o', linewidth=3, markersize=10, color='darkred')

# Add annotations for key events
plt.annotate('Pandemic\nLockdowns', xy=(2020, 5844), xytext=(2019.2, 5000),
             arrowprops=dict(facecolor='black', shrink=0.05, width=1.5, headwidth=8),
             fontsize=10, fontweight='bold', color='darkblue')

plt.annotate('Post-Pandemic\nSpike (11,108)', xy=(2023, 11108), xytext=(2021.5, 10500),
             arrowprops=dict(facecolor='black', shrink=0.05, width=1.5, headwidth=8),
             fontsize=10, fontweight='bold', color='darkred')

plt.title('Figure 1: Temporal Trend of Reported Crimes in Nairobi City Command (2018–2024)\nData Source: KNBS Economic Surveys', 
          fontsize=12, fontweight='bold', pad=15)
plt.xlabel('Year', fontsize=12)
plt.ylabel('Total Crimes Reported', fontsize=12)
plt.xticks(df_knbs['Year'], rotation=0)
plt.ylim(4000, 12000)
plt.tight_layout()

# Save the figure
plt.savefig('data/processed/fig1_nairobi_temporal_trend.png', dpi=300, bbox_inches='tight')
plt.show()
print("✅ Figure 1 saved to 'data/processed/fig1_nairobi_temporal_trend.png'")
```

```python
# 3. Load the NCRC Perception data
df_ncrc = pd.read_csv('data/processed/ncrc_nairobi_perception_gap.csv')

# Sort by Perception Gap to make the chart readable
df_ncrc_sorted = df_ncrc.sort_values(by='Perception_Gap', ascending=True)

# 4. Create the Perception Gap Chart
plt.figure(figsize=(10, 8))
# Color bars: Red if Nairobi perception is higher, Green if lower
colors = ['red' if x > 0 else 'green' for x in df_ncrc_sorted['Perception_Gap']]

plt.barh(df_ncrc_sorted['Crime_Category'], df_ncrc_sorted['Perception_Gap'], color=colors)
plt.axvline(0, color='black', linewidth=1)

plt.title('Figure 2: The "Perception Gap" - Nairobi vs. National Average\nPositive (Red) = Higher Perception in Nairobi', 
          fontsize=12, fontweight='bold', pad=15)
plt.xlabel('Difference in Percentage Points (%)', fontsize=12)
plt.ylabel('Crime Category', fontsize=12)
plt.grid(axis='x', linestyle='--', alpha=0.6)
plt.tight_layout()

# Save the figure
plt.savefig('data/processed/fig2_nairobi_perception_gap.png', dpi=300, bbox_inches='tight')
plt.show()
print("✅ Figure 2 saved to 'data/processed/fig2_nairobi_perception_gap.png'")
```
